All columns of type object (or string) → qualitative / categorical
All columns of type float64 or int64 → quantitative / numerical
Sometimes columns are read as object even though they contain numeric values (e.g., percentages, codes with dots or leading zeros) → these should be corrected or converted as needed for analysis.

```
import matplotlib.pyplot as plt
```

```
mean_age = clean_df['Age'].mean()
median_age = clean_df['Age'].median()
mode_age = clean_df['Age'].mode()[0]
```

```
print("Mean:", mean_age)
print("Median:", median_age)
print("Mode:", mode_age)
```

```
mean_age_crx = crx_df['0.1'].mean()
median_age_crx = crx_df['0.1'].median()
mode_age_crx = crx_df['0.1'].mode()[0]
print("Mean (crx):", mean_age_crx)
print("Median (crx):", median_age_crx)
print("Mode (crx):", mode_age_crx)
```

```
mean_age_sleep = sleep_df['Year'].mean()
median_age_sleep = sleep_df['Year'].median()
mode_age_sleep = sleep_df['Year'].mode()[0]
print("Mean (sleep):", mean_age_sleep)
print("Median (sleep):", median_age_sleep)
print("Mode (sleep):", mode_age_sleep)
```

```
mean_age_mental = mental_df['Depression (%)'].mean()
median_age_mental = mental_df['Depression (%)'].median()
mode_age_mental = mental_df['Depression (%)'].mode()[0]
print("Mean (mental):", mean_age_mental)
print("Median (mental):", median_age_mental)
print("Mode (mental):", mode_age_mental)
```

```
plt.plot(clean_df['Age'], clean_df['Debt'], marker='o', linestyle='-', color='blue')
plt.title("Age vs Debt (clean_dataset)")
plt.xlabel("Age")
plt.ylabel("Debt")
plt.show()
```

```
crx_df['1.25'] = crx_df['1.25'].astype(float)
crx_df['01'] = crx_df['01'].astype(int)

plt.plot(crx_df['1.25'], crx_df['01'], marker='o', linestyle='-', color='green')
plt.title("1.25 vs 01 (crx.csv)")
plt.xlabel("1.25")
plt.ylabel("01")
plt.show()
```
